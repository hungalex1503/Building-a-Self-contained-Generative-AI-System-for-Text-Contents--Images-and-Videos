```python
import torch
from diffusers import DiffusionPipeline, DPMSolverMultistepScheduler
from diffusers.utils import export_to_video
from IPython.display import Video, display

# 1. Cấu hình mô hình (Load Model)
print("⏳ Đang tải mô hình về (lần đầu sẽ mất khoảng 1-2 phút)...")
pipe = DiffusionPipeline.from_pretrained(
    "damo-vilab/text-to-video-ms-1.7b",
    torch_dtype=torch.float16,
    variant="fp16"
)

# Tối ưu hóa để chạy mượt trên Colab miễn phí (giảm ngốn RAM)
pipe.scheduler = DPMSolverMultistepScheduler.from_config(pipe.scheduler.config)
pipe.enable_model_cpu_offload()

# 2. Nhập nội dung video (Prompt)
prompt = "technology with modern cars"
negative_prompt = "low quality, distorted, watermark, blur"

print(f"🎬 Đang tạo video với mô tả: '{prompt}'...")

# 3. Chạy mô hình (Generate)
# num_frames: Số khung hình (càng cao video càng dài nhưng lâu)
video_frames = pipe(
    prompt,
    negative_prompt=negative_prompt,
    num_inference_steps=25,
    num_frames=16
).frames[0]

# 4. Lưu và hiển thị video
output_filename = "huggingface_video.mp4"
export_to_video(video_frames, output_filename, fps=8)

print("✅ Đã xong!")
display(Video(output_filename, embed=True, width=600))
```
